## 1. Use a pretrained model

This [DistilBERT model](https://huggingface.co/distilbert/distilbert-base-uncased-finetuned-sst-2-english) predicts whether a short English sentence is **POSITIVE** or **NEGATIVE**. It has no neutral label and can make mistakes.

Run the installation cell, then run the example. The model downloads on the first run. This public model does not require an API key.


```python
%pip -q install transformers torch
```

```python
# Install first: python -m pip install transformers torch
# In a Colab code cell, use: %pip -q install transformers torch

from transformers import pipeline

# Load a model that has already learned to classify English sentiment.
model = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english",
    device="cpu",
)

# Ask for a sentence, make a prediction, and show the result.
sentence = input("Write a short sentence in English: ")
result = model(sentence, truncation=True)[0]
print("Prediction:", result["label"])
```

### What happens in the code?

- `%pip install transformers torch` installs the libraries. `transformers` provides ready-to-use models and `torch` runs the neural network calculations.
- `from transformers import pipeline` imports a helper that connects a task, tokenizer, and model behind one simple interface.
- `pipeline("sentiment-analysis", ...)` creates a sentiment-analysis pipeline.
- `model=...` chooses a model already trained to classify English text as `POSITIVE` or `NEGATIVE`. It is downloaded the first time if it is not already cached.
- `device="cpu"` asks the model to run on the computer's processor, so a GPU is not needed.
- `input(...)` displays a question in the notebook and waits for you to type a sentence. The typed text is stored in `sentence`.
- `model(sentence, truncation=True)` sends the sentence through the tokenizer and model. The tokenizer turns text into numbers; the neural network uses learned parameters to estimate sentiment. `truncation=True` limits text that is longer than the model accepts.
- `[0]` takes the first result because the pipeline returns a list, even when we provide one sentence. The result contains a label and a score.
- `result["label"]` selects the label, and `print(...)` displays it. The code does not train or change the model.

### Try different sentences

Run the code cell again for each example:

- `I loved this movie!` will often be classified as `POSITIVE`.
- `I hated this movie!` will often be classified as `NEGATIVE`.
- `The movie was surprisingly good.` tests a positive opinion with different wording.
- `The food was cold, and the service was slow.` tests a negative opinion.
- `The movie started at seven.` is neutral in everyday language, but this model has no neutral label, so it must still choose `POSITIVE` or `NEGATIVE`.

The output also includes a score estimating how strongly the model favors its label. It is not a guarantee that the prediction is correct.

## 2. Ask an OpenAI model through the API

Create an API key in your [OpenAI account](https://platform.openai.com/api-keys). The account needs access to the selected model and available API quota; usage is billed through the API account. See the [official quickstart](https://developers.openai.com/api/docs/quickstart) and [GPT-4.1 mini model details](https://developers.openai.com/api/docs/models/gpt-4.1-mini).

Run the installation cell, then the example. Enter the key only in the hidden prompt provided by `getpass()`.

Each run sends one question to OpenAI. It does not keep a conversation history.


```python
%pip -q install openai
```

```python
# Install first: python -m pip install openai
# In a Colab code cell, use: %pip -q install openai

from getpass import getpass
from openai import OpenAI

# Enter your key privately; do not paste it into the notebook's code.
client = OpenAI(api_key=getpass("Enter your OpenAI API key: "))

# Ask a question and send it to an OpenAI model.
question = input("Ask the AI a question: ")
response = client.responses.create(
    model="gpt-4.1-mini",
    instructions="You are a friendly teacher. Answer in three short sentences.",
    input=question,
    max_output_tokens=200,
)

print(response.output_text)
```

### What happens from your code to the answer?

1. `%pip install openai` installs the OpenAI Python library, which helps Python make API requests.
2. `from getpass import getpass` imports a hidden text prompt. `from openai import OpenAI` imports the API client class.
3. `getpass(...)` asks for your API key without displaying the characters as you type. `OpenAI(api_key=...)` creates a client that can authenticate requests. Do not type the key into a code cell or share a notebook/output containing it.
4. `input(...)` asks you to type a question and saves it in `question`.
5. `client.responses.create(...)` builds and sends a request over the internet to OpenAI's API. The request includes:
   - `model="gpt-4.1-mini"`: which model should answer.
   - `instructions=...`: guidance about the answer's style and role. It is not Python code; it is text sent along with the question.
   - `input=question`: the question you typed.
   - `max_output_tokens=200`: an upper limit on the generated answer. A token is a piece of text and is not always a whole word.
6. On OpenAI's servers, the model processes the request and generates an answer piece by piece. The API sends the response back to the notebook.
7. `response.output_text` extracts the answer text, and `print(...)` displays it.

This example sends one question per run. It does not include earlier questions or answers, so the model does not automatically remember previous runs. Your question is sent outside your computer to the API. API access and quota are required, and usage may be billed to your OpenAI account.
